```python
SPAMFILE = r"../smsspamcollection/SMSSpamCollection"

labels = []
texts = []
with open(SPAMFILE, 'r', encoding='utf-8') as file:
    for line in file:
        l, t = line.strip().split('\t')
        if l == 'ham':
            labels.append(0)
        else:
            labels.append(1)
        texts.append(t)


x_train, x_test, y_train, y_test = train_test_split(texts, labels, test_size=0.2, random_state=20)


class_names = ['ham', 'spam']

vectorizer = sklearn.feature_extraction.text.TfidfVectorizer(lowercase=False)
#x_train_vectors = vectorizer.fit_transform(x_train)
#x_test_vectors = vectorizer.transform(x_test)
```

```python
from transformers import pipeline, AutoTokenizer, AutoModel
import torch
from tqdm import tqdm

tokenizer = AutoTokenizer.from_pretrained("bert-base-uncased")
model = AutoModel.from_pretrained("bert-base-uncased")

def get_bert_embeds(sentence):
    inputs = tokenizer(sentence, return_tensors='pt', return_attention_mask=True)
    with torch.no_grad():
        outputs = model(**inputs)
    return outputs.last_hidden_state[:, 0, :].squeeze(0).numpy()
    #return outputs.last_hidden_state.squeeze(0)

from sklearn.base import BaseEstimator, TransformerMixin

class BERTVectorizer(BaseEstimator, TransformerMixin):
    def __init__(self, model_name='bert-base-uncased', device=None):
        self.model_name = model_name
        self.device = device or ('cuda' if torch.cuda.is_available() else 'cpu')
        self.tokenizer = AutoTokenizer.from_pretrained(self.model_name)
        self.model = AutoModel.from_pretrained(self.model_name).to(self.device)

    def fit(self, X, y=None):
        return self  # no fitting needed

    def transform(self, X):
        vectors = []
        for sentence in X:
            inputs = self.tokenizer(sentence, return_tensors='pt', truncation=True,
                                    padding=True, max_length=128).to(self.device)
            with torch.no_grad():
                outputs = self.model(**inputs)
            # CLS token vector (shape: [1, hidden_dim])
            cls_vector = outputs.last_hidden_state[:, 0, :].cpu().numpy()
            vectors.append(cls_vector[0])
        return np.vstack(vectors)

train_vectors = [get_bert_embeds(sen) for sen in tqdm(x_train)]
print(np.shape(train_vectors))
test_vectors = [get_bert_embeds(sen) for sen in tqdm(x_test)]
```

```python
# Train modelRFRFRF - training a Random Forest just as an example

rf = RandomForestClassifier(random_state=42).fit(train_vectors, y_train)

mlp = MLPClassifier(solver='lbfgs', alpha=1e-5, hidden_layer_sizes=(50, 25), 
                         random_state=1).fit(train_vectors, y_train)

bert_vectorizer = BERTVectorizer()

modelRF = make_pipeline(bert_vectorizer, rf)
modelMLP = make_pipeline(bert_vectorizer, mlp)
```

#### F9.1 - Identity

```python
exp_list_lime = []
exp_list_ran = []
exp_list_dep = []
exp_list_con = []

# Choosing the first 30 instances of the test set
N = 30
# Choose how many repetitions we want
r = 10

# Iterate over instances to compute explanations for each instance multiple times
for i in range(N):
    # Initialize lists to hold repeated explanations for this instance
    instance_explanations_lime = []
    instance_explanations_ran = []
    instance_explanations_dep = []
    instance_explanations_con = []


    for j in range(r):
        # Compute LIME explanation
        exp_lime = lime_explainer.explain_instance(x_test[i], modelRF.predict_proba)
        # Extract the explanation values and append to the instance_explanations list
        instance_explanations_lime.append([v for _, v in exp_lime.as_list()])

        exp_ran = explainerRan.explain_instance(x_test[i], modelRF.predict_proba)
        instance_explanations_ran.append([v for _, v in exp_lime.as_list()])

        exp_dep = explainerDep.explain_instance(x_test[i], modelRF.predict_proba)
        instance_explanations_dep.append([v for _, v in exp_lime.as_list()])
        
        exp_con = explainerCon.explain_instance(x_test[i], modelRF.predict_proba)
        instance_explanations_con.append([v for _, v in exp_lime.as_list()])

    # Append the list of explanations for this instance to the main list
    exp_list_lime.append(instance_explanations_lime)    
    exp_list_ran.append(instance_explanations_ran)   
    exp_list_dep.append(instance_explanations_dep)    
    exp_list_con.append(instance_explanations_con)   
```

```python
# Spider plot for fixed, data-type, and black-box dependent scores
categories_fixed = ["f1_1", "f1_2", "f1_3", "f1_4", "f5_1", "f5_2", "f6_1", "f8_1", "f10"]
categories_dt = ["f2_1", "f2_2", "f2_3", "f2_4", "f4_1"]
categories_bb = ["f3", "f4_2", "f6_2", "f7_1", "f7_2", "f7_3", "f8_2", "f9_1", "f9_2", "f11"]

categories_excluding_total = [
    col for col in normalized_df.columns if col.startswith("f") and not col.endswith("_total")
]

categories_total_only = [col for col in normalized_df.columns if col.endswith("_total")]

def plot_spider(df, categories, title):
    angles = np.linspace(0, 2 * np.pi, len(categories), endpoint=False).tolist()
    angles += angles[:1]

    fig, ax = plt.subplots(figsize=(6, 6), subplot_kw=dict(polar=True))
    
    for _, row in df.iterrows():
        values = row[categories].values.flatten().tolist()
        values += values[:1]
        ax.plot(angles, values, label=row["Method"])
        ax.fill(angles, values, alpha=0.1)

    ax.set_yticks(np.linspace(0, 1, 6))
    ax.set_xticks(angles[:-1])
    ax.set_xticklabels(categories)
    ax.set_title(title)
    ax.legend(loc="upper right", bbox_to_anchor=(1.3, 1.1))

    # Save the plot as a PDF
    plt.savefig(f"generated/{title}.pdf", format="pdf", bbox_inches="tight")

    plt.show()
```
